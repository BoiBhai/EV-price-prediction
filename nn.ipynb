```python
import numpy as np
import pandas as pd
```

```python
train_processed = pd.read_csv('train_processed.csv', index_col='Unnamed: 0')
X_processed = train_processed.drop(columns='bin__Will_Buy_EV')
y_processed = train_processed['bin__Will_Buy_EV']
```

```python
from sklearn.model_selection import train_test_split

X, X_test, y, y_test = train_test_split(X_processed, y_processed, train_size=0.8)
```

```python
from tensorflow import keras
from tensorflow.keras import layers
```

```python
nn1 = keras.Sequential([
    layers.Input([20]),
    layers.Activation('sigmoid'),
    layers.Dense(32,),
    layers.Activation('sigmoid'),
    layers.Dense(32,),
    layers.Dropout(rate=0.3),
    layers.Activation('sigmoid'),
    layers.Dense(1),
])
```

```python
history = nn1.fit(
    X, y,
    validation_data=(X_test, y_test),
    batch_size=256,
    epochs=5,
    verbose=1,  # sets textual progress bars off
)
```
