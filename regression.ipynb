# Prelim

```python
import numpy as np
import pandas as pd
#import seaborn as sns
import matplotlib.pyplot as plt
```

```python
train_processed = pd.read_csv('train_processed.csv', index_col='id')
X_processed = train_processed.drop(columns='Unnamed: 0')
y_processed = pd.read_csv('train_y.csv', index_col='Unnamed: 0')
```

```python
X_processed
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix,precision_score, recall_score,f1_score,roc_curve, roc_auc_score

X, X_valid, y, y_valid = train_test_split(X_processed, y_processed, train_size=0.8)
```

```python
scaler = StandardScaler()

#X = scaler.fit_transform(X)
#X_valid = scaler.transform(X_valid)
```

# Logit Model

```python
y
```

```python
model = LogisticRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X_valid)
y_pred
```

```python
cm = confusion_matrix(y_valid, y_pred)
cm
```

```python
accuracy = accuracy_score(y_valid, y_pred)
precision = precision_score(y_valid, y_pred)
recall = recall_score(y_valid, y_pred)
f1 = f1_score(y_valid, y_pred)
```

```python
y_prob = model.predict_proba(X_valid)[:, 1]
fpr, tpr, thresholds = roc_curve(y_valid, y_prob)
roc_auc = roc_auc_score(y_valid, y_prob)
```

```python
from sklearn.metrics import RocCurveDisplay

# Directly plot from predictions
RocCurveDisplay.from_predictions(y_valid, y_prob)
plt.show()
```

# Regularization

```python
model = LogisticRegression(l1_ratio=0)
model.fit(X, y)
```

```python
y_pred = model.predict(X_valid)
y_pred
```

```python
cm = confusion_matrix(y_valid, y_pred)
cm
```

```python
accuracy = accuracy_score(y_valid, y_pred)
precision = precision_score(y_valid, y_pred)
recall = recall_score(y_valid, y_pred)
f1 = f1_score(y_valid, y_pred)
print(accuracy, precision, recall, f1)
```

```python
y_prob = model.predict_proba(X_valid)[:, 1]
fpr, tpr, thresholds = roc_curve(y_valid, y_prob)
roc_auc = roc_auc_score(y_valid, y_prob)
```

```python
RocCurveDisplay.from_predictions(y_valid, y_prob)
plt.show()
```

# Boosting

```python
from sklearn.ensemble import GradientBoostingClassifier

logitboost_model = GradientBoostingClassifier(verbose=True)
logitboost_model.fit(X,y)
```

```python
logitboost_model.score(X_valid, y_valid)
```

```python
from sklearn.metrics import RocCurveDisplay
y_prob = logitboost_model.predict_proba(X_valid)[:,1]
y_prob
```

```python
RocCurveDisplay.from_predictions(y_valid, y_prob)
plt.show()
```
